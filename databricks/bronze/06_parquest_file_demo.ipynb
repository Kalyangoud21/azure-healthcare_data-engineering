```python

# Read existing Bronze Delta table
df = spark.table(
    "dbw_healthcare_de_dev.bronze.patients_raw"
)

# Select a small sample for demonstration
df_sample = df.select(
    "patient_id",
    "first_name",
    "last_name",
    "gender",
    "date_of_birth",
    "state"
).limit(5)

# Destination in our own ADLS Gen2 account
output_path = (
    "abfss://bronze@examplehealthcarestorage.dfs.core.windows.net/"
    "parquet_demo/patients_sample"
)

# Write a single Parquet data file
(
    df_sample.coalesce(1)
    .write
    .format("parquet")
    .mode("error")
    .save(output_path)
)

print("Parquet sample created successfully")
print("Output location:", output_path)
```
