```sql
CREATE TABLE IF NOT EXISTS
dbw_healthcare_de_dev.silver.DimPatients (
    DimPatientID BIGINT GENERATED ALWAYS AS IDENTITY,
    patient_id STRING,
    first_name STRING,
    last_name STRING,
    gender STRING,
    date_of_birth DATE,
    state STRING,
    batch_id STRING,
    source_file STRING,
    ingestion_timestamp TIMESTAMP,
    record_hash STRING,
    effective_start_ts TIMESTAMP,
    effective_end_ts TIMESTAMP,
    is_current BOOLEAN
)
USING DELTA;

DESCRIBE TABLE dbw_healthcare_de_dev.silver.DimPatients


SELECT COUNT(*) AS record_count
FROM dbw_healthcare_de_dev.silver.DimPatients;
```
