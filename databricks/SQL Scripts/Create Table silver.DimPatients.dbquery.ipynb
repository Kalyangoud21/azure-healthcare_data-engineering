```sql

DROP TABLE IF EXISTS dbw_healthcare_de_dev.silver.DimPatients;

CREATE TABLE dbw_healthcare_de_dev.silver.DimPatients
(
    DimPatientID BIGINT GENERATED ALWAYS AS IDENTITY,
    PatientId STRING NOT NULL,
    FirstName STRING,
    LastName STRING,
    Gender STRING,
    DateOfBirth DATE,
    State STRING,

    BatchId STRING,
    SourceFile STRING,
    IngestionTimestamp TIMESTAMP,

    RecordHash STRING,
    RowStartDate TIMESTAMP,
    RowEndDate TIMESTAMP,
    IsActive BOOLEAN
)
USING DELTA
COMMENT 'Patient dimension supporting SCD Type 1 and Type 2';
```
